```python
import yaml
import os
from ultralytics import YOLO

# ================= 1. CẤU HÌNH DỮ LIỆU =================
data_config = {
    'path': r'D:\CPV301\Project\dataset full',
    'train': 'images/train',                   
    'val': 'images/val',                    
    'test': 'images/test',              
    'nc': 3,                              
    'names': ['Dao', 'Sung', 'Keo']       
}

# Tự động xuất ra file yaml
yaml_path = 'dataset_config.yaml'
with open(yaml_path, 'w', encoding='utf8') as f:
    yaml.dump(data_config, f, default_flow_style=False)

# ================= 2. BẮT ĐẦU HUẤN LUYỆN =================
if __name__ == '__main__':
    print("Đang tải thuật toán YOLOv8 bản Nano...")
    model = YOLO('yolov8n.pt') 
    
    print("Bắt đầu đưa vào lò huấn luyện...")
    results = model.train(
        data=yaml_path,
        epochs=50,          # Số vòng học (nếu máy chạy nhanh có thể tăng lên 100)
        imgsz=640,          # Kích thước ảnh chuẩn của YOLO
        batch=16,           # Số lượng ảnh học trong 1 lần (RTX 3050 Ti 4GB gánh tốt mức 16)
        device=0,           # BẮT BUỘC = 0 để sử dụng Card NVIDIA RTX 3050 Ti
        project='Do_An',    # Tên thư mục lưu kết quả
        name='Mo_Hinh_Cua_Toi' 
    )
    
    print("\nHUẤN LUYỆN HOÀN TẤT!")
    print("Mô hình thông minh nhất đã được lưu tại: Do_An/Mo_Hinh_Cua_Toi/weights/best.pt")
```

```python
import cv2
from ultralytics import YOLO

# 1. Tải "bộ não" bạn vừa huấn luyện xong ở Bước 3
# Đảm bảo đường dẫn này trỏ đúng tới file best.pt vừa sinh ra
duong_dan_model = r'D:\CPV301\Project\dataset full\runs\detect\Do_An\Mo_Hinh_Cua_Toi-2\weights\best.pt'
model = YOLO(duong_dan_model)

# 2. Mở Webcam (số 0 là camera mặc định của laptop)
cap = cv2.VideoCapture(0)

print("Đang mở Webcam... Nhấn phím 'q' trên bàn phím để tắt camera.")

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        print("Lỗi: Không đọc được tín hiệu từ Camera!")
        break

    # 3. Yêu cầu YOLO nhận diện trên từng khung hình video
    # Thông số conf=0.5 nghĩa là: AI phải chắc chắn độ tin cậy > 50% thì mới vẽ khung
    results = model.predict(
        source=frame, 
        conf=0.5, 
        device=0,  # Tiếp tục dùng card NVIDIA để xử lý nhanh
        show=True  # Tự động hiện cửa sổ màn hình video
    )

    # 4. Cài đặt phím tắt để thoát
    # Nhấn chữ 'q' (viết thường) để đóng cửa sổ
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Giải phóng bộ nhớ và đóng camera
cap.release()
cv2.destroyAllWindows()
```
